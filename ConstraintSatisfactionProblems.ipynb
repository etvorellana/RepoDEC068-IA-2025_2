Suponhamos que estamos analisando um mapa da Austrália que mostra cada um dos seus estados e territórios Recebemos a tarefa de colorir cada região de vermelho, verde ou azul, de forma que não haja duas regiões vizinhas da mesma cor. Para formular isso como um CSP, definimos as variáveis ​​como as regiões:
Um problema de satisfação de restrições consiste em três componentes, X,D e C: 

* $X$ é um conjunto de variáveis, $\{X_1,...,X_n\}$.
* $D$ é um conjunto de domínios, $\{D_1,...,D_n\}$, um para cada variável.
* $C$ é um conjunto de restrições que especifica combinações permitidas de valores.

Um domínio, $D_i$, consiste em um conjunto de valores permitidos, $\{v_1,...,v_k\}$, para a variável $X_i$. Por exemplo, uma variável booleana teria o domínio $\{\text{true, false}\}$. Variáveis ​​diferentes podem ter domínios diferentes de tamanhos diferentes. Cada restrição $C_j$ consiste em um par $⟨\text{escopo,rel}⟩$, onde escopo é uma tupla de variáveis ​​que participam da restrição e rel é uma relação que define os valores que essas variáveis ​​podem assumir. Uma relação pode ser representada como um conjunto explícito de todas as tuplas de valores que satisfazem a restrição, ou como uma função que pode calcular se uma tupla é membro da relação. Por exemplo, se $X_1$ e $X_2$ têm o domínio $\{1,2,3\}$, então a restrição que diz que $X_1$ deve ser maior que $X_2$ pode ser escrita como $⟨(X_1,X_2),\{(3,1), (3,2),(2,1)\}⟩$ ou como $⟨(X_1,X_2),X_1 > X_2⟩$.

Os CSPs lidam com atribuições de valores a variáveis, ${X_i = v_i , X_j = v_j , . . .}$. Uma atribuição que não viola nenhuma restrição é chamada de atribuição __consistente__ ou legal. Uma __atribuição completa__ é aquela em que cada variável recebe um valor, e uma __solução__ para um CSP é uma atribuição consistente e completa. Uma __atribuição parcial__ é aquela que deixa algumas variáveis ​​não atribuídas, e uma __solução parcial__ é uma atribuição parcial consistente. Resolver um CSP é um problema NP-completo em geral, embora existam importantes subclasses de CSPs que podem ser resolvidas de forma muito eficiente.

Para o problema do mapa da Australia temos que:

$X =\{WA,NT,Q,NSW,V,SA,T\}$

O domínio de cada variável é o conjunto $D_i = \{red, green, blue\}$. As restrições exigem que regiões vizinhas tenham cores distintas. Dado que existem nove locais onde as regiões fazem fronteira, existem nove restrições:

$C = \{SA \ne WA,SA \ne NT,SA \ne Q,SA \ne NSW,SA \ne V, WA \ne NT,NT \ne Q,Q \ne NSW, NSW  \ne V \}.$

Aqui estamos usando abreviações; $SA \ne WA$ é um atalho para $⟨(SA,WA),SA \ne WA⟩$, onde $SA \ne WA$ pode ser totalmente enumerado por sua vez como:

$\{(red, green), (red, blue), (green, red), (green, blue), (blue, red), (blue, green)\}$

```python
# uma solução mais simples
cores = ['red', 'green', 'blue']
restDiff = [(x,y) for x in cores for y in cores if x != y]
print(restDiff)
```

```python
# Variáveis 
# Vértices do grafo (Estados)
X = {'WA':{}, 'NT':{}, 'Q':{}, 'NSW':{}, 'V':{}, 'SA':{}, 'T':{}}

# Arestas do grafo (Fronteiras)
X['WA']['border'] = [('WA', 'NT'), ('WA', 'SA')]
X['NT']['border'] = [('NT', 'WA'), ('NT', 'SA'), ('NT', 'Q')]
X['Q']['border'] = [('Q', 'NT'), ('Q', 'SA'), ('Q', 'NSW')]
X['NSW']['border'] = [('NSW', 'Q'), ('NSW', 'SA'), ('NSW', 'V')]
X['V']['border'] = [('V', 'NSW'), ('V', 'SA')]
X['SA']['border'] = [('SA', 'WA'), ('SA', 'NT'), ('SA', 'Q'), ('SA', 'NSW'), ('SA', 'V')]
X['T']['border'] = []

print('X: ')
for x in X:
    X[x]['color'] = None
    print(x, ':', X[x])

# Domínios (Cores)
D = {x:cores for x in X}
print('D: ')
for d in D:
    print(d, ':', D[d])
```

```python
# Restrições
C = {}
for x in X:
    for y in X[x]['border']:
        C[(y)] = restDiff
print('C: ')
for c in C:
    print(c, ':', C[c])
```

```python
#An assignment that does not violate any constraints is called a consistent or legal assignment.
def isConsistent(X, D, x, value):
    if value not in D[x]:
        return False    
    if X[x]['border'] == []:
        return True
    for l, r in X[x]['border']:
        if X[r]['color'] == None:
            continue
        elif (X[r]['color'], value) in C[(l, r)]:
            return False
    return True

def assignments(X, D, x, value):
    if isConsistent(X, D, x, value):
        X[x]['color'] = value
    
```
